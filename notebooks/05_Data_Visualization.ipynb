```python
import pandas as pd
```

```python
df_customers_dummy = pd.DataFrame({
    "customer": ["A", "B", "C", "D", "E"],
    "age": [22, 35, 47, 63, 29],
    "income": [18000, 42000, 72000, 95000, 25000]
})

df_customers_dummy
```

```python
def age_group_label(age):
    if age < 30:
        return "Young"
    elif age < 63:
        return "Middle"
    else:
        return "Senior"

df_customers_dummy["age_group"] = df_customers_dummy["age"].apply(age_group_label)
df_customers_dummy
```

```python
def income_band(income):
    if income < 30000:
        return "low income"
    elif income < 60000:
        return "Middle income"
    else:
        return "High income"

df_customers_dummy ["income_band"] = df_customers_dummy ["income"].apply(income_band)
df_customers_dummy
```
